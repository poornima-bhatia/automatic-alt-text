```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'
}
def safe_request(url):
    try:
        # Use requests to check the status code first
        response = requests.get(url , headers = headers , verify=False , timeout=10)
        # driver.get(url)
        time.sleep(10)  # Wait for the page to load
        print(f"{url} => Page retrieved successfully!")
        return response.status_code
    except Exception as e:
        print(f"{url} => An error occurred: {e}")
        return None
def normalize_url(link_url):
    # Check if the URL is absolute
    if link_url.startswith('http://') or link_url.startswith('https://'):
        # Split the URL to remove everything after the domain
        normalized_url = link_url.split('/')[0] + '//' + link_url.split('/')[2]
        return normalized_url
    return link_url
# Load the Excel file

def process_sheet(sheet_name):
    # Load the Excel file
    df = pd.read_excel('sectors_data_x.xlsx', sheet_name=sheet_name)
    
    # Apply the normalize_url function to the 'link' column if it exists
    if 'link' in df.columns:
        df['link'] = df['link'].apply(normalize_url)
        response_status_codes = []
        statuses = []
        
        for link in df['link']:
            status_code = safe_request(link)
            response_status_codes.append(status_code)
            
            if status_code is not None:  # Check if the request was successful
                if 200 <= status_code < 300:
                    statuses.append("Page retrieved successfully!")
                elif status_code == 404:
                    statuses.append("Error: 404 Not Found")
                elif status_code == 503:
                    statuses.append("Error: 503 Service Unavailable")
                elif status_code >= 400:
                    statuses.append("Client Error")
                elif status_code >= 500:
                    statuses.append("Server Error")
            else:
                statuses.append("Request failed")
        
        df['response_status_code'] = response_status_codes
        df['status'] = statuses
    
    # Save the updated DataFrame back to the same Excel file
    with pd.ExcelWriter('sectors_data_x.xlsx', engine='openpyxl', mode='a' , if_sheet_exists='replace') as writer:
        df.to_excel(writer, sheet_name=sheet_name, index=False)
```

```python
excel_file = pd.ExcelFile('sectors_data_x.xlsx')
sheet_names = excel_file.sheet_names[1:30]
for sheet_name in sheet_names:
    process_sheet(sheet_name)
```
