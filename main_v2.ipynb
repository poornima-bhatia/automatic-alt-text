```python
import pandas as pd

# Load the Excel file
excel_file = pd.ExcelFile('sectors_data_x.xlsx')

# Initialize summary data
summary_report = []
links_with_no_alt_text = []  # To store links with images that have no alt text

# Process each sheet to collect summary data
for sheet_name in excel_file.sheet_names[1:30]:  # Adjust as needed
    df = pd.read_excel('sectors_data_x.xlsx', sheet_name=sheet_name)
    
    total_links = len(df)
    total_successful = df['response_status_code'].eq(200).sum()
    total_failed = total_links - total_successful
    total_images = df['total_images_src'].sum()
    total_missing_alt_text = df['total_images_alt_missing_text'].sum()
    
    # Check if total images equals total images with alt text
    for index, row in df.iterrows():
        if (row['response_status_code'] == 200) and (row['total_images_src'] == row['total_images_alt_missing_text']):
            links_with_no_alt_text.append(row['link'])
    
    # Append to summary report
    summary_report.append({
        'Sheet Name': sheet_name,
        'Total Links': total_links,
        'Total Successful (200 OK)': total_successful,
        'Total Failed': total_failed,
        'Total Images Checked': total_images,
        'Total Images Missing Alt Text': total_missing_alt_text,
    })

# Convert summary report to DataFrame
summary_df = pd.DataFrame(summary_report)

# Save the summary report to an Excel file
summary_df.to_excel('alt_text_summary_report.xlsx', index=False)

# Print the summary report
print("Summary Report")
print("====================")
print(summary_df)

# Print links where images equal alt text
print("\nLinks with Missing Alt Text:")
for link in links_with_no_alt_text:
    print(link)
```
